```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

days = 365
hours_per_day = 24
threshold = 35
```

```python
np.random.seed(42)

day = np.repeat(np.arange(days), hours_per_day)
hour = np.tile(np.arange(hours_per_day), days)

season_temp = 27 + 8 * np.sin(2 * np.pi * (day - 105) / days)
day_night_temp = 6 * np.sin(2 * np.pi * (hour - 9) / hours_per_day)
random_temp = np.random.normal(0, 1.2, days * hours_per_day)

temperatures = season_temp + day_night_temp + random_temp
temperatures[195 * 24:200 * 24] += 7

print("Number of readings:", len(temperatures))
print("First 10 temperatures:", np.round(temperatures[:10], 1))
```

```python
types = [np.float16, np.float32, np.float64]
results = []
daily_averages = {}

for data_type in types:
    readings = temperatures.astype(data_type)
    daily_avg = readings.reshape(days, hours_per_day).mean(
        axis=1,
        dtype=np.float64
    )

    alert_days = np.zeros(days, dtype=bool)

    for i in range(days - 2):
        if daily_avg[i] > threshold and daily_avg[i + 1] > threshold and daily_avg[i + 2] > threshold:
            alert_days[i:i + 3] = True

    daily_averages[data_type.__name__] = daily_avg

    results.append({
        "Data type": data_type.__name__,
        "Memory (KB)": readings.nbytes / 1024,
        "Days above 35°C": sum(daily_avg > threshold),
        "Alert days": sum(alert_days)
    })

results_table = pd.DataFrame(results)
display(results_table)
```

```python
accuracy_results = []

for data_type in types:
    converted = temperatures.astype(data_type).astype(np.float64)
    difference = np.max(np.abs(temperatures - converted))

    accuracy_results.append({
        "Data type": data_type.__name__,
        "Largest difference (°C)": difference
    })

accuracy_table = pd.DataFrame(accuracy_results)
display(accuracy_table)
```

```python
daily_avg = daily_averages["float32"]
days_list = np.arange(1, days + 1)
alert_days = np.zeros(days, dtype=bool)

for i in range(days - 2):
    if daily_avg[i] > threshold and daily_avg[i + 1] > threshold and daily_avg[i + 2] > threshold:
        alert_days[i:i + 3] = True

plt.figure(figsize=(12, 5))
plt.plot(days_list, daily_avg, label="Daily average")
plt.axhline(threshold, color="orange", linestyle="--", label="35°C threshold")
plt.scatter(days_list[alert_days], daily_avg[alert_days], color="red", label="Alert days")

plt.xlabel("Day of the year")
plt.ylabel("Temperature (°C)")
plt.title("Simulated Temperatures and Heatwave Alerts")
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.bar(results_table["Data type"], results_table["Memory (KB)"])
plt.title("Memory used")
plt.ylabel("KB")

plt.subplot(1, 2, 2)
plt.bar(accuracy_table["Data type"], accuracy_table["Largest difference (°C)"])
plt.title("Largest rounding difference")
plt.ylabel("°C")

plt.tight_layout()
plt.show()
```
